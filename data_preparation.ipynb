```python
import pandas as pd
import requests
```

```python
BASE_URL = "https://data-charts-api.hexlet.app"

START = "2023-03-01"
END = "2023-09-01"
```

```python
visits_response = requests.get(
    f"{BASE_URL}/visits",
    params={"begin": START, "end": END},
    timeout=30
)

visits_response.raise_for_status()

visits_data = visits_response.json()
```

```python
registrations_response = requests.get(
    f"{BASE_URL}/registrations",
    params={"begin": START, "end": END},
    timeout=30
)

registrations_response.raise_for_status()

registrations_data = registrations_response.json()
```

```python
print("Посещения:")
print(type(visits_data))
print("Количество записей:", len(visits_data))

print("\nРегистрации:")
print(type(registrations_data))
print("Количество записей:", len(registrations_data))
```

```python
visits_data[:3]
registrations_data[:3]
```

```python
visits = pd.DataFrame(visits_data)
registrations = pd.DataFrame(registrations_data)
display(visits.head())
display(registrations.head())
```

```python
print("Размер таблицы посещений:", visits.shape)
print("Размер таблицы регистраций:", registrations.shape)
visits.info()
registrations.info()
```

```python
visits["datetime"] = pd.to_datetime(visits["datetime"], errors="coerce")
registrations["datetime"] = pd.to_datetime(
    registrations["datetime"],
    errors="coerce"
)
```
